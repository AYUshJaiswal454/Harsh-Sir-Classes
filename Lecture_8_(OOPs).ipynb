```python
class complex:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __add__(self,other):
    a = self.x + other.x
    b = self.y + other.y

    result = complex(a, b)

    return result


  def display(self):
    print(self.x, '+i', self.y)

    a = complex(2, 3)
    b = complex(3, 4)

    c = a + b

    print(c.x, c.y)
```
