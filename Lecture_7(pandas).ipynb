```python
import pandas as pd

df1 = pd.DataFrame({
    'id': [1,2,3,4],
    'name': ['Raja','Rani','Rahul','Rani'],
    'age': [23,24,25,26]
})

df2 = pd.DataFrame({
    'id': [3,4,5,6],
    'name': ['Rahul','Rani','Mohan','Sohan'],
    'age': [23,24,34,43]
})


# df3 = pd.merge(df1,df2,on='id',how='outer')
# print(df3)

df3 = pd.merge(df1,df2,on='id',how='inner')
print(df3)
```

```python
df1 = pd.DataFrame({
    'id': [1,2,3,4],
    'name': ['Raja','Rani','Rahul','Rani'],
    'age': [23,24,25,26]
})

df2 = pd.DataFrame({
    'id': [3,4,5,6],
    'name': ['Rahul','Rani','Mohan','Sohan'],
    'age': [23,24,34,43]
})

# df3 = pd.merge(df1,df2,on='id',how='left')
# print(df3)

df3 = pd.merge(df1,df2,on='id',how='right')
print(df3)
```
