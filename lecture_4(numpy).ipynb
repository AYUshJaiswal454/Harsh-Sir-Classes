```python
# Random Number Generation

a = np.random.randint(0,9,2)
print(a)
```

```python

# Broadcasting

a = np.array([1,2,3,4])

b = np.array([9,3,4,6])
print(a*2)
print(a+b)
print(a*b)
print(a/b)
print((2*a)+b)
```

```python
# Array into Matrix

a = np.random.randint(2, 15, 6)
print(a)

print(a.reshape(2,3))


print(a.reshape(3,2))
```

```python
# Transpose

a = np.random.randint(2, 15, 6)
print(a)
b = a.reshape(2,3)
print(b)
print(b.transpose())
```

```python
# Concatenate in 1-D array

a = np.random.randint(2, 15, 6)
print(a)
b = np.random.randint(2, 15, 6)
print(b)


print(np.concatenate((a,b)))


```
