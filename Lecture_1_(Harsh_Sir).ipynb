```python
a = input()

n=len(a)
j=n-1
ans=""


for i in range(0,n,1) :
  if(a[i]==a[j]) :
    ans="yes"
    j=j-1
  else :
    ans="no";
    break;
if ans=="yes" :
  print("Palindrome")
else :
  print(-1)
```
